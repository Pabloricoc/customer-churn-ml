### Customer identifier validation

`customerID` should uniquely identify each customer. Verifying its uniqueness helps confirm that each row represents a different customer.

```python
n_rows = len(df)
n_unique_customers = df["customerID"].nunique()
duplicated_customer_ids = df["customerID"].duplicated().sum()

print(f"Number of rows: {n_rows}")
print(f"Unique customer IDs: {n_unique_customers}")
print(f"Duplicated customer IDs: {duplicated_customer_ids}")
```

## Data quality findings

The initial data quality assessment identified the following findings:

- The dataset contains 7,043 observations and 21 variables.
- Each row represents a unique customer, confirmed by the uniqueness of `customerID`.
- `customerID` is an identifier and should not be used as a predictive feature.
- `TotalCharges` is stored as an object although it conceptually represents a numerical quantity.
- Some `TotalCharges` values contain blank strings that are not detected by `isna()`.
- The problematic `TotalCharges` observations correspond to customers with `tenure = 0`.
- No completely duplicated rows were detected.
- `SeniorCitizen` is stored as an integer but should conceptually be treated as a binary categorical variable.
- The target variable `Churn` is moderately imbalanced, with the majority of customers belonging to the non-churn class.
- No transformations have been applied to the raw data at this stage.
